```python

import re
import html
import pandas as pd

def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Decode HTML entities
    text = html.unescape(text)

    # Normalize backtick apostrophes
    text = text.replace("`", "'")

    # Lowercase
    text = text.lower()

    # Remove Markdown links, keeping visible text
    text = re.sub(r'\[([^\]]+)\]\([^)]+\)', r'\1', text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Remove HTML tags
    text = re.sub(r'<[^>]*>', ' ', text)

    # Remove @mentions
    text = re.sub(r'@\w+', ' ', text)

    # Keep hashtag text
    text = re.sub(r'#(\w+)', r'\1', text)

    # Expand common contractions
    text = re.sub(r"\bcan't\b", "cannot", text)
    text = re.sub(r"\bwon't\b", "will not", text)
    text = re.sub(r"n't\b", " not", text)
    text = re.sub(r"'re\b", " are", text)
    text = re.sub(r"'ve\b", " have", text)
    text = re.sub(r"'ll\b", " will", text)
    text = re.sub(r"'m\b", " am", text)
    text = re.sub(r"'d\b", " would", text)

    # Handle common emoticons before removing punctuation
    text = re.sub(r'[:;=8][-^]?[)(/DPp]', ' EMOTICON ', text)
    text = re.sub(r'<3', ' LOVE ', text)

    # Remove censored words such as ****
    text = re.sub(r'\*+', ' ', text)

    # Keep only letters and whitespace
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# Apply to your DataFrame
df['clean_text'] = df['text'].apply(clean_text)

# Check the output
print(df[['text', 'clean_text', 'sentiment']].head(15))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
from sklearn.model_selection import train_test_split

X = df['clean_text']
y = df['sentiment']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.5,stratify=y)

print('Training shape:',X_train.shape)
print('Testing shape:',X_test.shape)

```

```python
# converting into word count vectors
cv = CountVectorizer()

X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)

print("Training shape:", X_train_cv.shape)
print("Testing shape:", X_test_cv.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score,classification_report,confusion_matrix)

```

```python
# training

lr = LogisticRegression()
lr.fit(X_train_cv,y_train)

```

```python
y_pred_lr = lr.predict(X_test_cv)
```

```python
print("accuracy_score:",accuracy_score(y_test,y_pred_lr))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

```python
word_tokenize(X_train.iloc[0])
```

```python
sentences = [word_tokenize(tweet) for tweet in X_train]

print("Number of training tweets:", len(sentences))
print("First tokenized tweet:", sentences[0])
```

```python
w2v = Word2Vec(sentences=sentences,vector_size=100,window=5,min_count=2,workers=2,sg=1,epochs=5,seed=42)
```

```python
w2v.wv['watching']
```

```python
print("Vocabulary size:", len(w2v.wv))

print("\nVector for 'watching':")
print(w2v.wv["watching"])

print("\nVector shape:", w2v.wv["watching"].shape)
```

```python
tweet_lengths = [len(tokens) for tokens in sentences]


print("Shortest tweet:", min(tweet_lengths))
print("Longest tweet:", max(tweet_lengths))
print("Average tweet length:", sum(tweet_lengths) / len(tweet_lengths))

print("\nLength percentiles:")
print("50th percentile:", np.percentile(tweet_lengths, 50))
print("90th percentile:", np.percentile(tweet_lengths, 90))
print("95th percentile:", np.percentile(tweet_lengths, 95))
print("99th percentile:", np.percentile(tweet_lengths, 99))
```

```python
max_length = 25
```

```python
X_train_vectors = [
    [w2v.wv[word] for word in word_tokenize(tweet)
     if word in w2v.wv]
    for tweet in X_train
]
```

```python
print("First tweet:", X_train.iloc[0])
print("Number of word vectors:", len(X_train_vectors[0]))
print("Shape of first word vector:", X_train_vectors[0][0].shape)
```

```python
X_test_vectors = [
    [w2v.wv[word] for word in word_tokenize(tweet)
     if word in w2v.wv]
    for tweet in X_test
]
```

```python
print("Number of test tweets:", len(X_test_vectors))
print("First test tweet:", X_test.iloc[0])
print("Number of word vectors:", len(X_test_vectors[0]))
```

```python
max_length = 25
vector_size = w2v.vector_size

X_train_padded = np.zeros(
    (len(X_train_vectors), max_length, vector_size),
    dtype=np.float32
)

for i, tweet_vectors in enumerate(X_train_vectors):
    tweet_vectors = np.asarray(tweet_vectors, dtype=np.float32)
    length = min(len(tweet_vectors), max_length)

    if length > 0:
        X_train_padded[i, :length] = tweet_vectors[:length]

print("Training shape:", X_train_padded.shape)
print("Tweets with no known word vectors:", sum(len(v) == 0 for v in X_train_vectors))
```

```python
X_test_padded = np.zeros(
    (len(X_test_vectors), max_length, vector_size),
    dtype=np.float32
)

for i, tweet_vectors in enumerate(X_test_vectors):
    tweet_vectors = np.asarray(tweet_vectors, dtype=np.float32)
    length = min(len(tweet_vectors), max_length)

    if length > 0:
        X_test_padded[i, :length] = tweet_vectors[:length]

print("Training shape:", X_train_padded.shape)
print("Testing shape:", X_test_padded.shape)
```

```python
model_lstm = tf.keras.Sequential([tf.keras.layers.Masking(mask_value = 0.0,input_shape = (max_length,vector_size)),
                     tf.keras.layers.LSTM(64),
                     tf.keras.layers.Dense(32,activation="relu"),
                     tf.keras.layers.Dropout(0.3),
                     tf.keras.layers.Dense(3,activation='softmax')])

```
